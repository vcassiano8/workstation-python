```python
import matplotlib.pyplot as plt
from scipy import stats

dados = np.array([10, 12, 13, 15, 18, 20, 21, 23, 25, 27])

print("=== ANÁLISE CIENTÍFICA ===")
print("Média:", np.mean(dados))
print("Mediana:", np.median(dados))
print("Desvio padrão:", np.std(dados))

df = pd.DataFrame({"valor": dados})

print("\nDataFrame:")
display(df)

resultado = stats.describe(dados)

print("\nEstatísticas SciPy:")
print(resultado)

plt.figure(figsize=(8, 5))
plt.plot(dados, marker="o")
plt.title("Dados experimentais simulados")
plt.xlabel("Observação")
plt.ylabel("Valor")
plt.grid(True)
plt.show()
```
